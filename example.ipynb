```python
%matplotlib inline
from sup import sup
from sup import settings as sup_defaults
```

There is a docstring.

```python
sup?
```

How to call with a dictionary:

```python
s = sup_defaults.copy()
# make changes to s
sup(s)
```

How to call with a settings file:

```python
sup(setfile="settings.yaml")
```

How to call with default settings plus an object selection from a built-in catalog:

```python
sup(setfile="messier_openclusters M11 M26")
```

The above calling styles can be mixed, with anything specified by `setfile` overriding `settings`.

```python
s = sup_defaults.copy()
# make changes to s
sup(s, "more_settings.yaml")
```

If the built-in catalogs are needed and are not in the working directory, their path must be provided, as in:

```python
sup(setfile="messier_openclusters M11 M26", thispath="/path/to/installation")
# or, sup(setfile="/path/to/installation/messier_openclusters M11 M26")
```

You can make and use your own catalogs in the same format. Explicitly setting the path to built-in catalogs is not necessary when running from the command line.
